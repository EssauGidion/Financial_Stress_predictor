```python
TARGET = "liquidity_stress_next_30d"
y = train[TARGET]
y.head()
```

```python
ID_COL = "ID"

feature_cols = [
    c for c in train.columns
    if c not in [ID_COL, TARGET]
]

print("Number of features:", len(feature_cols))
```

```python
X = train[feature_cols]

X.head()
```

```python
cat_cols = X.select_dtypes(include="object").columns.tolist()

print("Number of categorical features:", len(cat_cols))
print(cat_cols)
```

```python
for col in cat_cols:
    print(f"\n{col}:")
    print(X[col].unique())
```

```python
X_encoded = X.copy()

X_encoded.head()
```

```python
X_encoded = pd.get_dummies(
    X_encoded,
    columns=cat_cols,
    drop_first=True
)

print("Features after encoding:", X_encoded.shape[1])
```

```python
X_encoded.dtypes.value_counts()
```

```python
X_encoded.isnull().sum().sum()
```

```python
from sklearn.model_selection import train_test_split

X_train, X_val, y_train, y_val = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_val:", X_val.shape)
print("y_train:", y_train.shape)
print("y_val:", y_val.shape)
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_val_scaled = scaler.transform(X_val)

print("X_train_scaled:", X_train_scaled.shape)
print("X_val_scaled:", X_val_scaled.shape)
```

```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=100,
    class_weight="balanced",
    solver="liblinear"
)

model.fit(X_train_scaled, y_train)

print("Model training completed!")
```

```python
val_pred_prob = model.predict_proba(X_val_scaled)[:, 1]

print(val_pred_prob[:10])
```

```python
from sklearn.metrics import roc_auc_score

roc_auc = roc_auc_score(y_val, val_pred_prob)

print("Validation ROC-AUC:", roc_auc)
```

```python
val_pred_label = (val_pred_prob >= 0.5).astype(int)

print(val_pred_label[:20])
```

```python
from sklearn.metrics import classification_report

print(classification_report(y_val, val_pred_label))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_val, val_pred_label)

print(cm)
```

```python
from sklearn.inspection import permutation_importance
```

```python
result = permutation_importance(
    model,
    X_val_scaled,
    y_val,
    n_repeats=3,
    random_state=42,
    n_jobs=-1
)

print("Permutation importance completed!")
```

```python
importances = pd.Series(
    result.importances_mean,
    index=X_encoded.columns
)

print(importances.sort_values(ascending=False).head(10))
```

```python
from sklearn.model_selection import train_test_split

X_train_model, X_calib, y_train_model, y_calib = train_test_split(
    X_train,
    y_train,
    test_size=0.25,
    random_state=42,
    stratify=y_train
)

print("Model training:", X_train_model.shape)
print("Calibration:", X_calib.shape)
print("Final validation:", X_val.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf_calibration_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=12,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

rf_calibration_model.fit(
    X_train_model,
    y_train_model
)

print("Random Forest training completed.")
```

```python
from sklearn.calibration import CalibratedClassifierCV

calibrated_rf = CalibratedClassifierCV(
    rf_calibration_model,
    method="sigmoid",
    cv=5
)

calibrated_rf.fit(X_calib, y_calib)

print("Calibration completed.")
```

```python
from sklearn.metrics import log_loss, roc_auc_score

calibrated_predictions = calibrated_rf.predict_proba(X_val)[:, 1]

calibrated_logloss = log_loss(
    y_val,
    calibrated_predictions
)

calibrated_auc = roc_auc_score(
    y_val,
    calibrated_predictions
)

print("Calibrated Random Forest")
print("Log Loss:", calibrated_logloss)
print("ROC-AUC:", calibrated_auc)
```

```python
from sklearn.metrics import log_loss, roc_auc_score

# Original RF v2 predictions
rf_v2_predictions = rf_model_v2.predict_proba(X_val)[:, 1]

rf_v2_logloss = log_loss(y_val, rf_v2_predictions)
rf_v2_auc = roc_auc_score(y_val, rf_v2_predictions)

# Calibrated RF predictions
calibrated_logloss = log_loss(y_val, calibrated_predictions)
calibrated_auc = roc_auc_score(y_val, calibrated_predictions)

# Competition-style weighted score
rf_v2_score = (
    0.60 * rf_v2_logloss +
    0.40 * (1 - rf_v2_auc)
)

calibrated_score = (
    0.60 * calibrated_logloss +
    0.40 * (1 - calibrated_auc)
)

print("RF v2")
print("Log Loss:", rf_v2_logloss)
print("ROC-AUC:", rf_v2_auc)
print("Combined score:", rf_v2_score)

print("\nCalibrated RF")
print("Log Loss:", calibrated_logloss)
print("ROC-AUC:", calibrated_auc)
print("Combined score:", calibrated_score)
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.calibration import CalibratedClassifierCV

final_rf = RandomForestClassifier(
    n_estimators=200,
    max_depth=12,
    random_state=42,
    class_weight="balanced",
    n_jobs=-1
)

final_calibrated_rf = CalibratedClassifierCV(
    final_rf,
    method="sigmoid",
    cv=5
)

final_calibrated_rf.fit(X_encoded, y)

print("Final calibrated Random Forest trained.")
```
